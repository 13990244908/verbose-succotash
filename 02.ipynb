# 第二部分：调用智谱 GLM 大模型 API

这一部分是本次作业的**重点**：用自己的 Python 代码把大模型调通。

思路只有四步：

1. 从 `.env` 里读出 API Key（不把 Key 写进代码）
2. 用 Key 创建客户端
3. 把你的问题（提示词）发给模型
4. 取出模型返回的文字

如果还没配置 Key，下面单元格会给出提示而不是报错。

**先装依赖**：`pip install -r requirements.txt`。
注意 SDK 的**安装包名是 `zai-sdk`**，但导入时写 `from zai import ZhipuAiClient`。

````python
# 11. 检查环境是否就绪
# 依赖没装先执行：pip install zai-sdk python-dotenv

import os
from dotenv import load_dotenv

load_dotenv()  # 读取当前目录的 .env 文件

api_key = os.environ.get("ZHIPU_API_KEY")

if api_key:
    print("API Key 已就绪，长度:", len(api_key))
else:
    print("未找到 ZHIPU_API_KEY")
    print("请把 .env.example 复制为 .env，并填入你在 open.bigmodel.cn 申请的 Key")
````

````python
# 12. 调用 GLM 模型（一次最简单的对话）

from zai import ZhipuAiClient

if not api_key:
    print("跳过：请先配置 API Key")
else:
    # 初始化客户端
    client = ZhipuAiClient(api_key=api_key)

    # 创建聊天完成请求
    response = client.chat.completions.create(
        model="glm-5.3",  # 模型名；若提示无权限或不存在，可换成 "glm-4.6"
        messages=[
            {"role": "system", "content": "您是一个有用的AI助手。"},
            {"role": "user", "content": "您好，请介绍一下自己。"},
        ],
        temperature=0.6,
    )

    # 获取回复
    print(response.choices[0].message.content)
````

````python
# 13. 封装成函数：以后只要传入提示词就能用

def zhipuAI(提示词, temperature=0.6):
    """向智谱 GLM 发送一条提示词，返回模型的回复文字。"""

    import os
    from dotenv import load_dotenv
    from zai import ZhipuAiClient

    load_dotenv()  # 读取当前目录 .env

    api_key = os.environ.get("ZHIPU_API_KEY")
    if not api_key:
        raise RuntimeError("未找到 ZHIPU_API_KEY，请检查 .env 文件")

    # 初始化客户端
    client = ZhipuAiClient(api_key=api_key)

    # 创建聊天完成请求
    response = client.chat.completions.create(
        model="glm-5.3",  # 模型名；若提示无权限或不存在，可换成 "glm-4.6"
        messages=[
            {"role": "system", "content": "您是一个有用的AI助手。"},
            {"role": "user", "content": f"{提示词}"},
        ],
        temperature=temperature,
    )

    # 返回回复内容
    return response.choices[0].message.content

print("zhipuAI 函数已定义，之后只要传提示词就能调用")
````

````python
# 14. 作业任务：用自己封装的函数，让 GLM 介绍你的专业

if not api_key:
    print("跳过：请先配置 API Key")
else:
    print(zhipuAI("您好，请介绍一下西北民族大学广告学专业。"))
````
